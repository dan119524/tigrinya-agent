```python
from langchain.agents import create_agent
from langchain.tools import tool

@tool
def check_length(text: str) -> str:
    """Count the words in a translation."""
    count = len(text.split())
    return f"The translation has {count} words."

agent = create_agent(
    model="anthropic:claude-sonnet-5-5",
    tools=[check_length],
    system_prompt="Translate English to Tigrinya.Use check_length to count the words.",
)

result = agent.invoke({"messages": [{"role": "user", "content": "Translate: Good morning, how are you?"}]})
print(result["messages"][-1].content)
```

```python
from pydantic import BaseModel

class Translation(BaseModel):
    english: str
    to_man: str
    to_woman: str

agent = create_agent(
    model="anthropic:claude-sonnet-5-5",
    tools=[check_length],
    system_prompt="Translate English to Tigrinya. Never use Amharic. Use check_length to count the words.",
    response_format=Translation,
)

result = agent.invoke({"messages": [{"role": "user", "content": "Translate: Where is the market?"}]})
print(result["structured_response"])
```

```python
from pydantic import BaseModel
from langchain.agents.structured_output import ProviderStrategy   # NEW

class Translation(BaseModel):
    english: str
    to_man: str
    to_woman: str

agent = create_agent(
    model="anthropic:claude-sonnet-5-5",
    tools=[check_length],
    system_prompt="Translate English to Tigrinya. Never use Amharic. Use check_length to count the words.",
    response_format=ProviderStrategy(Translation),   # CHANGED
)

result = agent.invoke({"messages": [{"role": "user", "content": "Translate: Where is the market?"}]})
print(result["structured_response"])
```

```python
from dataclasses import dataclass
from langchain.agents.middleware import dynamic_prompt, ModelRequest

@dataclass
class UserContext:
    formal: bool

@dynamic_prompt
def choose_tone(request: ModelRequest) -> str:
    base = "Translate English to Tigrinya. Never use Amharic. Explain in English."
    if request.runtime.context.formal:
        return base + " Use a formal, respectful tone."
    return base + " Use a casual, friendly tone."

agent = create_agent(
    model="anthropic:claude-sonnet-5-5",
    middleware=[choose_tone],
    context_schema=UserContext,
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "Translate: How are you?"}]},
    context=UserContext(formal=True),
)
print(result["messages"][-1].content)
```

```python
from langchain.agents.middleware import HumanInTheLoopMiddleware
from langgraph.types import Command

agent = create_agent(
    model="anthropic:claude-sonnet-5-5",
    tools=[check_length],
    system_prompt="Translate English to Tigrinya. Never use Amharic. Explain in English. Use check_length to count the words.",
    checkpointer=InMemorySaver(),
    middleware=[HumanInTheLoopMiddleware(
        interrupt_on={"check_length": {"allowed_decisions": ["approve", "reject"]}}
    )],
)

config = {"configurable": {"thread_id": "2"}}
result = agent.invoke({"messages": [{"role": "user", "content": "Translate: Where are you going?"}]}, config)
print(result.get("__interrupt__"))
```

```python
result = agent.invoke(Command(resume={"decisions": [{"type": "approve"}]}), config)
print(result["messages"][-1].content)
```

```python
result = agent.invoke(Command(resume={"decisions": [{"type": "approve"}, {"type": "approve"}]}), config)
print(result["messages"][-1].content)
```

```python
result = agent.invoke(Command(resume={"decisions": [{"type": "approve"}, {"type": "approve"}]}), config)
print(result["messages"][-1].content)
```

```python
config = {"configurable": {"thread_id": "3"}}
result = agent.invoke({"messages": [{"role": "user", "content": "Translate: Where are you going?"}]}, config)

while "__interrupt__" in result:
    count = len(result["__interrupt__"][0].value["action_requests"])
    print(f"Approving {count} tool call(s)")
    decisions = [{"type": "approve"}] * count
    result = agent.invoke(Command(resume={"decisions": decisions}), config)

print(result["messages"][-1].content)
```
